```python
import pandas as pd
import zipfile
import pickle

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix, classification_report
```

```python
dataset = pd.read_csv('D:/All Vs Code/Machine learning/Veil/dataset/super_sms_dataset.csv',encoding='latin1')

#we dont need preprocessing cause the data is already plenty refined
#having said that we culd have dropped useless columns from the csv itself.
```

```python
x_train, x_test, y_train, y_test = train_test_split(dataset["SMSes"],dataset["Labels"])\

#by default it does a 75 percent 25 percent train and test data split which is completely fine
```

```python
tfidf_vect = TfidfVectorizer(use_idf=True, lowercase=True, ngram_range=(1, 2),sublinear_tf=True,min_df=2)
x_train_tfidf = tfidf_vect.fit_transform(x_train)
x_train_tfidf.shape

#only convert xtrain cause thats whats my model supposed to predict not the answers , i.e, 0s and 1s
```

```python
model = MultinomialNB(alpha=0.5)
model.fit(x_train_tfidf, y_train)
```

```python
x_test_idf = tfidf_vect.transform(x_test)
predictions = model.predict(x_test_idf)
```

```python
confusion_matrix(y_test,predictions)
```

```python
print(classification_report(y_test, predictions))
```

```python
with open("model.pkl","wb") as model_file:
    pickle.dump(model,model_file)

with open("tfidf-vect.pkl","wb") as tfidf_vect_file:
    pickle.dump(tfidf_vect, tfidf_vect_file)
```
